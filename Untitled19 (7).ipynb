# Section 6: Data Transformation (Wide to Long Format)

The temporal datasets are initially stored in wide format, where each product is represented as a separate column. While this structure is suitable for spreadsheet-based analysis, it is not appropriate for graph-based machine learning.

Therefore, each temporal dataset is transformed into long format, where every row represents the operational value of one product on one date. This standardized structure enables efficient merging of multiple datasets, simplifies feature engineering, and supports scalable preprocessing using PySpark.

```python
# ============================================================
# START PYSPARK SESSION
# ============================================================

from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = SparkSession.builder \
    .appName("SupplyChainRiskBigDataProcessing") \
    .getOrCreate()

print("Spark Session Started Successfully")
print("Spark Version:", spark.version)
```

```python
# ============================================================
# CONVERT EXISTING OPERATIONAL DATASETS TO SPARK DATAFRAMES
# ============================================================

unit_sales_spark = spark.createDataFrame(unit_sales)
unit_production_spark = spark.createDataFrame(unit_production)
unit_factory_spark = spark.createDataFrame(unit_factory)
unit_delivery_spark = spark.createDataFrame(unit_delivery)

print("Spark DataFrames created successfully.")

print("\nSales Schema:")
unit_sales_spark.printSchema()

print("\nProduction Schema:")
unit_production_spark.printSchema()

print("\nFactory Issue Schema:")
unit_factory_spark.printSchema()

print("\nDelivery Schema:")
unit_delivery_spark.printSchema()
```

```python
# ============================================================
# PySpark Function: Convert Wide Format to Long Format
# ============================================================

from pyspark.sql import functions as F

def wide_to_long_spark(df, value_name):

    id_col = "Date"

    value_cols = [
        c for c in df.columns
        if c != id_col
    ]

    # Create an array of structs instead of using stack()
    # This safely handles columns such as POP001L12P.1
    array_expr = F.array(*[
        F.struct(
            F.lit(c).alias("Product"),
            F.col(f"`{c}`").cast("double").alias(value_name)
        )
        for c in value_cols
    ])

    return (
        df
        .select(
            F.col(id_col),
            F.explode(array_expr).alias("temp")
        )
        .select(
            F.col(id_col),
            F.col("temp.Product").alias("Product"),
            F.col(f"temp.{value_name}").alias(value_name)
        )
        .filter(F.col(value_name).isNotNull())
    )


print("wide_to_long_spark function created successfully.")
```

```python
# ============================================================
# Convert Sales from Wide Format to Long Format using PySpark
# ============================================================

sales_long_spark = wide_to_long_spark(
    unit_sales_spark,
    "Sales"
)

print("Sales wide-to-long conversion completed.")

print("Rows:", sales_long_spark.count())
print("Columns:", len(sales_long_spark.columns))

sales_long_spark.printSchema()

sales_long_spark.show(10, truncate=False)
```

```python
# ============================================================
# Create Unified Product-Level Supply Chain Dataset
# ============================================================

supply_chain_spark = (
    sales_long_spark
    .join(
        production_long_spark,
        on=["Date", "Product"],
        how="inner"
    )
    .join(
        factory_long_spark,
        on=["Date", "Product"],
        how="inner"
    )
    .join(
        delivery_long_spark,
        on=["Date", "Product"],
        how="inner"
    )
)

print("Unified supply chain dataset created successfully.")

print("Rows:", supply_chain_spark.count())
print("Columns:", len(supply_chain_spark.columns))

supply_chain_spark.printSchema()

supply_chain_spark.show(10, truncate=False)
```

# Section 9: Business Feature Engineering

The statistical features created in the previous section summarize the operational behavior of each product. However, supply chain risk is influenced not only by absolute quantities but also by operational efficiency, demand fluctuations, production gaps, and delivery performance.

In this section, several business-oriented features are engineered to better represent the operational health of each product. These features will later be used for computing the product risk score and will also serve as node features for the Graph Neural Network.

```python
eps = 1e-6
product_features["ProductionEfficiency"] = (
    product_features["TotalProduction"] /
    (product_features["TotalSales"] + eps)
)
product_features["DeliveryEfficiency"] = (
    product_features["TotalDelivery"] /
    (product_features["TotalSales"] + eps)
)
product_features["FactoryIssueRatio"] = (
    product_features["TotalFactoryIssue"] /
    (product_features["TotalProduction"] + eps)
)
product_features["SalesGap"] = (
    product_features["TotalSales"] -
    product_features["TotalDelivery"]
)
product_features["ProductionGap"] = (
    product_features["TotalProduction"] -
    product_features["TotalSales"]
)
product_features["SalesCV"] = (
    product_features["StdSales"] /
    (product_features["AvgSales"] + eps)
)
product_features["ProductionCV"] = (
    product_features["StdProduction"] /
    (product_features["AvgProduction"] + eps)
)
```

# Section 10: Risk Score Generation

The original dataset does not contain a risk label for each product. Therefore, a composite Risk Score is generated using operational performance indicators.

The generated Risk Score represents the overall operational health of each product and will later serve as the prediction target for the Graph Neural Network.

Products with poor production efficiency, high factory issues, unstable demand, and large operational gaps will receive higher risk scores.

# Section 11: Risk Score Generation using PCA

The dataset does not contain predefined risk labels for products.

Instead of manually assigning weights to operational metrics, Principal Component Analysis (PCA) is used to combine multiple business indicators into a single composite Risk Score.

This approach allows the data itself to determine the importance of each feature, making the generated risk score objective and scientifically justifiable.

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=1)
risk_score = pca.fit_transform(risk_scaled)
risk_score[:5]
```

```python
from sklearn.preprocessing import MinMaxScaler
risk_scaler = MinMaxScaler(feature_range=(0,100))
risk_score = risk_scaler.fit_transform(risk_score)
risk_score[:10]
```

```python
product_features["RiskScore"] = risk_score
product_features.head()
```

```python
def risk_level(score):
    if score < 25:
        return "Low"
    elif score < 50:
        return "Medium"
    elif score < 75:
        return "High"
    else:
        return "Critical"
product_features["RiskLevel"] = product_features["RiskScore"].apply(risk_level)
```

# Part B: rectified methodology
1. Chronological windows: features from the first 150 days, target from the remaining 71 days.
2. Observed target: delivery shortfall rate in the target window, cut into tertiles by rank, for products with at least MIN_FUTURE_SALES of demand there.
3. One row per product: POP001L12P.1 is merged into POP001L12P, so rows match the 40 graph nodes in the same order.
4. Baselines: majority class, persistence, Random Forest and MLP with graph degree; GCN with permuted-node and single-relation controls.
5. Evaluation: 5-fold x 10-repeat stratified cross-validation, scaler fitted on training nodes only, Wilcoxon tests on repeat means.

## Chronological Data Split & Feature/Target Window Engineering

To eliminate data leakage, we separate our temporal dataset chronologically.
* **Feature Window:** First 150 days (used to compute node features and construct graph relationships).
* **Target Window:** Remaining 71 days (used to compute actual observed risk events, such as delivery shortfalls relative to sales demand).

# ## Section 24: Rolling-window panel (many labelled examples)
# One snapshot every 7 days. Features come from the previous 28 days, the target from the next 14 days.

```python

LOOKBACK, HORIZON, STRIDE = 28, 14, 7
MIN_FUTURE_SALES_H = 50.0                          # units ordered in the horizon below which a shortfall rate is not defined
GAP = int(np.ceil(HORIZON / STRIDE))               # snapshots dropped between train and test so target windows never overlap

dates_all = sorted(md["Date"].unique()); T = len(dates_all)
prods = list(product_features["Product"]); NP = len(prods)
piv = {c: md.pivot(index="Date", columns="Product", values=c).reindex(index=dates_all, columns=prods).fillna(0.0).values
       for c in ["Sales", "Production", "FactoryIssue", "Delivery"]}          # each array is (days, products)
starts = list(range(LOOKBACK, T - HORIZON + 1, STRIDE)); n_snap = len(starts)

PF = ["TotalSales", "TotalProduction", "TotalFactoryIssue", "TotalDelivery", "SalesCV", "ProductionCV",
      "ProductionEfficiency", "DeliveryEfficiency", "FactoryIssueRatio", "SalesGap", "ProductionGap",
      "Shortfall7d", "ShortfallLookback"]
def snap_features(t):
    sl = slice(t - LOOKBACK, t)
    S, P, Fi, D = (piv[c][sl] for c in ("Sales", "Production", "FactoryIssue", "Delivery"))
    S_t, P_t, F_t, D_t = S.sum(0), P.sum(0), Fi.sum(0), D.sum(0)
    cv = lambda A: A.std(0) / (A.mean(0) + eps)
    s7, d7 = S[-7:].sum(0), D[-7:].sum(0)
    return np.column_stack([S_t, P_t, F_t, D_t, cv(S), cv(P), P_t / (S_t + eps), D_t / (S_t + eps), F_t / (P_t + eps),
                            S_t - D_t, P_t - S_t, (s7 - d7) / (s7 + eps), (S_t - D_t) / (S_t + eps)])

Xraw = np.stack([snap_features(t) for t in starts])                       # (snapshots, products, 13)
futS = np.stack([piv["Sales"][t:t + HORIZON].sum(0) for t in starts])
futD = np.stack([piv["Delivery"][t:t + HORIZON].sum(0) for t in starts])
short = (futS - futD) / (futS + eps)

Xt = Xraw.copy()
for c in [0, 1, 2, 3, 9, 10]:                                              # heavy-tailed counts and gaps: signed log
    Xt[..., c] = np.sign(Xt[..., c]) * np.log1p(np.abs(Xt[..., c]))
Xt[..., [6, 7, 8]] = np.clip(Xt[..., [6, 7, 8]], -10, 10)
Xt[..., [4, 5]] = np.clip(Xt[..., [4, 5]], 0, 10)
Xt[..., [11, 12]] = np.clip(Xt[..., [11, 12]], -5, 1)

deg = np.array([dict(G.degree())[p] for p in prods], dtype=float)
Xt = np.concatenate([Xt, np.tile(deg, (n_snap, 1))[..., None]], axis=2)    # add GraphDegree
FEATS_P = PF + ["GraphDegree"]

# neighbour-mean features: the best non-neural way to give a tabular model the graph
ei_np = edge_index.cpu().numpy() if hasattr(edge_index, "cpu") else np.asarray(edge_index)
A = np.zeros((NP, NP)); np.add.at(A, (ei_np[0], ei_np[1]), 1.0)
A = A / np.maximum(A.sum(1, keepdims=True), 1.0)
NBp = np.stack([A @ Xt[s, :, :len(PF)] for s in range(n_snap)]).reshape(n_snap * NP, len(PF))

Xp = Xt.reshape(n_snap * NP, -1)
Short = short.reshape(-1); FutS = futS.reshape(-1)
snap_id = np.repeat(np.arange(n_snap), NP); node_id = np.tile(np.arange(NP), n_snap)
elig = FutS >= MIN_FUTURE_SALES_H
print(f"Snapshots: {n_snap} | rows: {len(Short)} | labelled rows: {int(elig.sum())} ({100*elig.mean():.0f}%) | features: {Xp.shape[1]}")
print("Future shortfall quantiles over labelled rows (10, 33, 50, 67, 90%):", np.quantile(Short[elig], [.1, .33, .5, .67, .9]).round(4))
```

 #Section 25: Models and temporal cross-validation
# Binary target: High = future shortfall above the training-period 67th percentile (thresholds never see test data).

```python

from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import balanced_accuracy_score, f1_score, roc_auc_score, accuracy_score
from scipy.stats import spearmanr

N_FOLDS = 3
test_len = max(3, n_snap // (N_FOLDS + 1))
folds = []
for k in range(N_FOLDS):
    te_start = n_snap - (N_FOLDS - k) * test_len
    te_s = np.arange(te_start, te_start + test_len)
    tr_s = np.arange(0, te_start - GAP)
    folds.append((np.where(np.isin(snap_id, tr_s) & elig)[0], np.where(np.isin(snap_id, te_s) & elig)[0]))
    print(f"fold {k}: train snapshots 0-{tr_s[-1]} ({len(folds[-1][0])} rows) | test snapshots {te_s[0]}-{te_s[-1]} ({len(folds[-1][1])} rows)")

COL_LB = FEATS_P.index("ShortfallLookback")

def p_majority(tr, te, yb, seed):
    return np.zeros(len(te)), np.zeros(len(te), dtype=int)

def p_persistence(tr, te, yb, seed):
    thr = np.quantile(Xp[tr, COL_LB], 2 / 3)               # "next window looks like the last window"
    s = Xp[te, COL_LB]
    return s, (s > thr).astype(int)

def _rf(X):
    def fn(tr, te, yb, seed):
        m = RandomForestClassifier(n_estimators=300, min_samples_leaf=3, class_weight="balanced_subsample",
                                   random_state=seed, n_jobs=-1).fit(X[tr], yb[tr])
        s = m.predict_proba(X[te])[:, 1]
        return s, (s > 0.5).astype(int)
    return fn
p_rf = _rf(Xp)
p_rf_nb = _rf(np.hstack([Xp, NBp]))

def fold_metrics(y, s, pred, sh):
    return {"BalAcc": balanced_accuracy_score(y, pred), "MacroF1": f1_score(y, pred, average="macro", zero_division=0),
            "AUC": roc_auc_score(y, s) if len(np.unique(y)) == 2 and np.ptp(s) > 0 else 0.5,
            "Spearman": spearmanr(s, sh)[0] if np.ptp(s) > 0 else 0.0, "Acc": accuracy_score(y, pred)}
```

# Section 25 (continued): MLP and GCN on the snapshot graphs
# Every snapshot is its own copy of the 40-node graph; copies are disjoint, so nothing leaks between snapshots.

```python

import torch, torch.nn as nn, torch.nn.functional as F
from torch_geometric.nn import GCNConv

assert int(edge_index.max()) < NP
def union_edges(ei):
    return torch.cat([ei + s * NP for s in range(n_snap)], dim=1)
EI_UNION = union_edges(edge_index)

class PMLP(nn.Module):
    def __init__(self, d):
        super().__init__(); self.a, self.b, self.o = nn.Linear(d, 32), nn.Linear(32, 16), nn.Linear(16, 2)
    def forward(self, x, ei=None):
        h = F.dropout(F.relu(self.a(x)), 0.2, self.training)
        return self.o(F.relu(self.b(h)))

class PGCN(nn.Module):
    def __init__(self, d):
        super().__init__(); self.c1, self.c2, self.o = GCNConv(d, 32), GCNConv(32, 16), nn.Linear(16, 2)
    def forward(self, x, ei):
        h = F.dropout(F.relu(self.c1(x, ei)), 0.2, self.training)
        return self.o(F.relu(self.c2(h, ei)))

def train_nn(make, tr, yb, seed, graph, epochs=200):
    torch.manual_seed(seed)
    sc = StandardScaler().fit(Xp[tr])                                       # scaler sees training rows only
    Xs = torch.tensor(sc.transform(Xp), dtype=torch.float)
    y = torch.tensor(yb, dtype=torch.long); trt = torch.tensor(tr, dtype=torch.long)
    p1 = float(yb[tr].mean()); w = torch.tensor([1 / (1 - p1), 1 / p1], dtype=torch.float); w = w / w.sum() * 2
    m = make(Xs.shape[1]); opt = torch.optim.Adam(m.parameters(), lr=0.01, weight_decay=5e-4)
    for _ in range(epochs):
        m.train(); opt.zero_grad()
        F.cross_entropy(m(Xs, graph)[trt], y[trt], weight=w).backward(); opt.step()
    m.eval()
    return m, sc, Xs

def _nn(make, graph_fn, seeds=(0, 1, 2)):
    def fn(tr, te, yb, seed):
        s = np.zeros(len(te))
        for sd in seeds:
            m, sc, Xs = train_nn(make, tr, yb, 100 * seed + sd, graph_fn(100 * seed + sd))
            with torch.no_grad():
                s += torch.softmax(m(Xs, graph_fn(100 * seed + sd)), 1)[:, 1].numpy()[te] / len(seeds)
        return s, (s > 0.5).astype(int)
    return fn
p_mlp = _nn(PMLP, lambda sd: None)
p_gcn = _nn(PGCN, lambda sd: EI_UNION)
p_gcn_perm = _nn(PGCN, lambda sd: union_edges(torch.tensor(np.random.RandomState(sd).permutation(NP))[edge_index]))
def _rel(rel): return _nn(PGCN, lambda sd: union_edges(rel_edge_index[rel]))
```


#  Section 31: Features v2. Within-product change and delivery backlog
# Persistence is below chance (late deliveries are caught up later), so the model needs backlog and change-versus-own-history features.

```python

cum = np.cumsum(piv["Sales"] - piv["Delivery"], axis=0)                       # cumulative undelivered sales per product
bl = np.stack([cum[t - 1] for t in starts])                                   # backlog at the start of each snapshot
wk_sales = np.stack([piv["Sales"][t - LOOKBACK:t].sum(0) for t in starts]) / (LOOKBACK / 7)
BLW = np.clip(bl / (wk_sales + eps), -10, 10)                                 # backlog measured in weeks of sales

def _short(a, b):
    S, D = piv["Sales"][a:b].sum(0), piv["Delivery"][a:b].sum(0)
    return (S - D) / (S + eps)
prev_short = np.stack([_short(t - 2 * LOOKBACK, t - LOOKBACK) if t >= 2 * LOOKBACK else Xraw[i, :, 12]
                       for i, t in enumerate(starts)])
prev_short = np.clip(prev_short, -5, 1)
trend = Xt[..., 12] - prev_short                                              # is the shortfall getting better or worse?

def vs_own_past(A):                                                           # value minus the mean of the product's EARLIER snapshots
    out = np.zeros_like(A)
    for s in range(1, A.shape[0]):
        out[s] = A[s] - A[:s].mean(0)
    return out
DEV_COLS = [0, 3, 6, 7, 11, 12]        # TotalSales, TotalDelivery, ProductionEfficiency, DeliveryEfficiency, Shortfall7d, ShortfallLookback
DEV = np.stack([vs_own_past(Xt[..., c]) for c in DEV_COLS], axis=2)

Xt2 = np.concatenate([Xt, BLW[..., None], prev_short[..., None], trend[..., None], DEV], axis=2)
FEATS2 = FEATS_P + ["BacklogWeeks", "PrevShortfall", "ShortfallTrend"] + ["d_" + FEATS_P[c] for c in DEV_COLS]
Xp2 = Xt2.reshape(n_snap * NP, -1)
print("Features v2:", Xp2.shape[1], "| new:", FEATS2[len(FEATS_P):])
```


#  Section 32: GCN-only models v2 (less smoothing, validation-based epochs and threshold)
# One GCN layer with self-loop weight 2, a skip from the product's own features, DropEdge and dropout.
# With `layers=0` the same class is a graph-free network, which gives a like-for-like control.

```python


import torch, torch.nn as nn, torch.nn.functional as F
from torch_geometric.nn import GCNConv
from sklearn.metrics import roc_auc_score

class GCNv2(nn.Module):
    def __init__(self, d, hid=32, layers=1, p_drop=0.3, p_edge=0.3):
        super().__init__()
        self.skip = nn.Linear(d, hid)
        self.convs = nn.ModuleList([GCNConv(d if i == 0 else hid, hid, improved=True) for i in range(layers)])
        self.o = nn.Linear(hid, 2); self.p_drop, self.p_edge = p_drop, p_edge
    def forward(self, x, ei=None):
        if self.training and ei is not None and self.p_edge > 0:                 # DropEdge
            ei = ei[:, torch.rand(ei.size(1)) > self.p_edge]
        h = F.relu(self.skip(x)); h0 = h
        for i, c in enumerate(self.convs):
            h = h + F.relu(c(x if i == 0 else h, ei))
        return self.o(F.dropout(h, self.p_drop, self.training))

def _train(make, X, rows, yb, seed, graph, n_ep, score_rows=None):
    torch.manual_seed(seed)
    sc = StandardScaler().fit(X[rows])                                          # scaler sees training rows only
    Xs = torch.tensor(sc.transform(X), dtype=torch.float)
    y = torch.tensor(yb, dtype=torch.long); rt = torch.tensor(rows, dtype=torch.long)
    p1 = float(yb[rows].mean()); w = torch.tensor([1 / (1 - p1), 1 / p1], dtype=torch.float); w = w / w.sum() * 2
    m = make(Xs.shape[1]); opt = torch.optim.Adam(m.parameters(), lr=0.01, weight_decay=5e-4)
    hist = []
    for _ in range(n_ep):
        m.train(); opt.zero_grad()
        F.cross_entropy(m(Xs, graph)[rt], y[rt], weight=w).backward(); opt.step()
        if score_rows is not None:
            m.eval()
            with torch.no_grad():
                hist.append(torch.softmax(m(Xs, graph), 1)[:, 1].numpy()[score_rows])
    m.eval()
    with torch.no_grad():
        full = torch.softmax(m(Xs, graph), 1)[:, 1].numpy()
    return full, hist

def _nn2(make, graph_fn, X, seeds=(0, 1, 2), max_ep=300):
    def fn(tr, te, yb, fold):
        last = snap_id[tr].max()
        inner = tr[snap_id[tr] <= last - 1 - GAP]                                # inner training period
        val = tr[snap_id[tr] >= last - 1]                                        # validation = last 2 training snapshots (never the test fold)
        s_te = np.zeros(len(te)); thr_all = []
        for sd in seeds:
            seed = 100 * fold + sd; g = graph_fn(seed)
            _, hist = _train(make, X, inner, yb, seed, g, max_ep, score_rows=val)
            yv = yb[val]
            if len(np.unique(yv)) == 2:
                auc = np.array([roc_auc_score(yv, h) for h in hist])
                auc = np.convolve(auc, np.ones(5) / 5, mode="same")             # smooth, then pick the best epoch
                E = int(np.argmax(auc)) + 1
                cand = np.quantile(hist[E - 1], np.linspace(0.3, 0.8, 26))
                bal = [balanced_accuracy_score(yv, (hist[E - 1] > c).astype(int)) for c in cand]
                thr = float(cand[int(np.argmax(bal))])
            else:
                E, thr = max_ep // 2, 0.5
            full, _ = _train(make, X, tr, yb, seed, g, E)                         # retrain on the whole training period for E epochs
            s_te += full[te] / len(seeds); thr_all.append(thr)
        t = float(np.mean(thr_all))
        return s_te, (s_te > t).astype(int)
    return fn

EI_PS = torch.cat([rel_edge_index["Plant"], rel_edge_index["ProductSubGroup"]], dim=1)
perm_graph = lambda sd: union_edges(torch.tensor(np.random.RandomState(sd).permutation(NP))[edge_index])
mk_gcn = lambda d: GCNv2(d, layers=1)
mk_nog = lambda d: GCNv2(d, layers=0)

```

# ## Section 34: Features v3 (lagged snapshot) and a feature-similarity graph
# The lag features give the model the last snapshot's state and the change since then.
# The similarity graph links each product to its 5 most similar products in the SAME snapshot (features only, never labels).
# %%

```python


import torch
LAGN = ["TotalSales", "DeliveryEfficiency", "Shortfall7d", "ShortfallLookback", "BacklogWeeks"]
LAGC = [FEATS2.index(n) for n in LAGN]
Xl = np.zeros((n_snap, NP, len(LAGC)))
Xl[1:] = Xt2[:-1][..., LAGC]                                   # previous snapshot (zeros for the first one)
Xt3 = np.concatenate([Xt2, Xl, Xt2[..., LAGC] - Xl], axis=2)   # current, lag, and change since the last snapshot
FEATS3 = FEATS2 + ["lag_" + n for n in LAGN] + ["chg_" + n for n in LAGN]
Xp3 = Xt3.reshape(n_snap * NP, -1)
print("Features v3:", Xp3.shape[1])

K_SIM = 5
SIMC = [FEATS2.index(n) for n in ["TotalSales", "DeliveryEfficiency", "Shortfall7d", "BacklogWeeks"]]
def knn_edges(Z, k):
    D2 = ((Z[:, None, :] - Z[None, :, :]) ** 2).sum(-1); np.fill_diagonal(D2, np.inf)
    nn_ = np.argsort(D2, axis=1)[:, :k]
    s_, d_ = np.repeat(np.arange(len(Z)), k), nn_.reshape(-1)
    return np.unique(np.concatenate([np.stack([s_, d_]), np.stack([d_, s_])], axis=1), axis=1)
eis = []
for s in range(n_snap):
    Z = Xt2[s][:, SIMC]; Z = (Z - Z.mean(0)) / (Z.std(0) + 1e-6)
    eis.append(torch.tensor(knn_edges(Z, K_SIM) + s * NP, dtype=torch.long))
EI_KNN = torch.cat(eis, dim=1)
print("Similarity graph edges per snapshot (directed): %.0f vs %d in the relation graph" % (EI_KNN.size(1) / n_snap, edge_index.size(1)))

```

# ## Section 35: Per-relation GCN (one convolution per relation type)
# %%

```python

import torch.nn as nn, torch.nn.functional as F
from torch_geometric.nn import GCNConv
RELS = ["Plant", "Storage", "ProductGroup", "ProductSubGroup"]

class GCNrel(nn.Module):
    """A skip from the product's own features plus one GCN convolution per relation type."""
    def __init__(self, d, hid=32, p_drop=0.3, p_edge=0.3):
        super().__init__()
        self.skip = nn.Linear(d, hid)
        self.convs = nn.ModuleList([GCNConv(d, hid, improved=True) for _ in RELS])
        self.o = nn.Linear(hid, 2); self.p_drop, self.p_edge = p_drop, p_edge
    def forward(self, x, eis):
        h = F.relu(self.skip(x))
        for c, ei in zip(self.convs, eis):
            if self.training and self.p_edge > 0:
                ei = ei[:, torch.rand(ei.size(1)) > self.p_edge]
            h = h + F.relu(c(x, ei)) / len(self.convs)
        return self.o(F.dropout(h, self.p_drop, self.training))

EIS_REL = [union_edges(rel_edge_index[r]) for r in RELS]
mk_rel = lambda d: GCNrel(d)
```

# Runs the same models on four defensible targets and prints every result, so the choice is disclosed and not cherry-picked.
# shortfall = share of ordered sales not delivered in the horizon; backlog = open orders (cumulative sales minus deliveries) at the end of the horizon, in weeks of sales.


```python

from sklearn.metrics import accuracy_score
Xuse = Xp3 if "Xp3" in globals() else Xp2                      # Section 34 features if you ran Part E, else Section 31 features
GAP_14 = GAP

def build_target(mode):
    H = 28 if mode.endswith("28") else 14
    fS = np.full((n_snap, NP), np.nan); fD = np.full((n_snap, NP), np.nan); BLn = np.full((n_snap, NP), np.nan)
    cumS = np.cumsum(piv["Sales"] - piv["Delivery"], axis=0)
    for i, t in enumerate(starts):
        if t + H <= T:
            fS[i] = piv["Sales"][t:t + H].sum(0); fD[i] = piv["Delivery"][t:t + H].sum(0); BLn[i] = cumS[t + H - 1]
    ok = (~np.isnan(fS)) & (np.nan_to_num(fS) >= MIN_FUTURE_SALES_H * H / 14)
    wk_ = np.stack([piv["Sales"][t - LOOKBACK:t].sum(0) for t in starts]) / (LOOKBACK / 7)
    if mode.startswith("shortfall"):
        tgt, pers = (fS - fD) / (fS + eps), Xt[..., COL_LB]
    else:
        tgt, pers = np.clip(BLn / (wk_ + eps), -10, 10), BLW
    return np.nan_to_num(tgt).reshape(-1), ok.reshape(-1), pers.reshape(-1), int(np.ceil(H / STRIDE))

def make_folds(ok, gap, n_folds=3):
    test_len = max(3, n_snap // (n_folds + 1)); out = []
    for k in range(n_folds):
        te_start = n_snap - (n_folds - k) * test_len
        tr_s, te_s = np.arange(0, te_start - gap), np.arange(te_start, te_start + test_len)
        out.append((np.where(np.isin(snap_id, tr_s) & ok)[0], np.where(np.isin(snap_id, te_s) & ok)[0]))
    return out

def _nn2_safe(make, graph_fn, X, seeds=(0, 1, 2), max_ep=300):
    def fn(tr, te, yb, fold):
        last = snap_id[tr].max() if len(tr) > 0 else 0
        inner = tr[snap_id[tr] <= last - 1 - GAP]
        val = tr[snap_id[tr] >= last - 1]
        if len(inner) == 0:
            inner = tr
            val = np.array([])
        s_te = np.zeros(len(te)); thr_all = []
        for sd in seeds:
            seed = 100 * fold + sd; g = graph_fn(seed)
            if len(val) > 0:
                _, hist = _train(make, X, inner, yb, seed, g, max_ep, score_rows=val)
                yv = yb[val]
                if len(np.unique(yv)) == 2:
                    auc = np.array([roc_auc_score(yv, h) for h in hist])
                    auc = np.convolve(auc, np.ones(5) / 5, mode="same")
                    E = int(np.argmax(auc)) + 1
                    cand = np.quantile(hist[E - 1], np.linspace(0.3, 0.8, 26))
                    bal = [balanced_accuracy_score(yv, (hist[E - 1] > c).astype(int)) for c in cand]
                    thr = float(cand[int(np.argmax(bal))])
                else:
                    E, thr = max_ep // 2, 0.5
            else:
                E, thr = max_ep // 2, 0.5
            full, _ = _train(make, X, tr, yb, seed, g, E)
            s_te += full[te] / len(seeds); thr_all.append(thr)
        t = float(np.mean(thr_all))
        return s_te, (s_te > t).astype(int)
    return fn

table = []
for mode in ["shortfall14", "shortfall28", "backlog14", "backlog28"]:
    tgt, ok, pers, gap = build_target(mode); GAP = gap
    res = {}
    for k, (tr, te) in enumerate(make_folds(ok, gap)):
        if len(tr) == 0:
            continue
        yb = (tgt > np.quantile(tgt[tr], 2 / 3)).astype(int)
        sp, tp = pers[te], np.quantile(pers[tr], 2 / 3)
        preds = {"Majority class": (np.zeros(len(te)), np.zeros(len(te), dtype=int)),
                 "Persistence": (sp, (sp > tp).astype(int)),
                 "Random Forest": _rf(Xuse)(tr, te, yb, k),
                 "No-graph network": _nn2_safe(mk_nog, lambda sd: None, Xuse)(tr, te, yb, k),
                 "GCN": _nn2_safe(mk_gcn, lambda sd: EI_UNION, Xuse)(tr, te, yb, k)}
        for name, (s, p) in preds.items():
            res.setdefault(name, []).append(fold_metrics(yb[te], s, p, tgt[te]))
    for name, ms in res.items():
        table.append({"Target": mode, "Model": name, "Acc": np.mean([m["Acc"] for m in ms]),
                      "BalAcc": np.mean([m["BalAcc"] for m in ms]), "AUC": np.mean([m["AUC"] for m in ms]),
                      "labelled rows": int(ok.sum())})
    print("done:", mode)
GAP = GAP_14
study = pd.DataFrame(table)
display(study.pivot(index="Model", columns="Target", values="Acc").round(3).rename_axis("ACCURACY"))
display(study.pivot(index="Model", columns="Target", values="BalAcc").round(3).rename_axis("BALANCED ACCURACY"))
display(study.pivot(index="Model", columns="Target", values="AUC").round(3).rename_axis("AUC"))
study.round(4).to_csv("Target_Study.csv", index=False)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import precision_score, recall_score
from xgboost import XGBClassifier

Xfin, FEATS_FIN = Xp3, FEATS3
col_deg = FEATS_FIN.index("GraphDegree")
Xnd = np.delete(Xfin, col_deg, axis=1)                       # the same features without the degree column

def metrics_full(y, s, pred, sh):
    m = fold_metrics(y, s, pred, sh)
    m["Precision"] = precision_score(y, pred, zero_division=0)   # precision and recall of the High class
    m["Recall"] = recall_score(y, pred, zero_division=0)
    return m

def _lr(X):
    def fn(tr, te, yb, seed):
        sc = StandardScaler().fit(X[tr])
        m = LogisticRegression(C=0.5, class_weight="balanced", max_iter=2000).fit(sc.transform(X[tr]), yb[tr])
        s = m.predict_proba(sc.transform(X[te]))[:, 1]
        return s, (s > 0.5).astype(int)
    return fn

def _xgb(X):
    def fn(tr, te, yb, seed):
        pos = float(yb[tr].mean())
        m = XGBClassifier(n_estimators=300, max_depth=3, learning_rate=0.05, subsample=0.8, colsample_bytree=0.8,
                          scale_pos_weight=(1 - pos) / pos, eval_metric="logloss", random_state=seed, n_jobs=-1).fit(X[tr], yb[tr])
        s = m.predict_proba(X[te])[:, 1]
        return s, (s > 0.5).astype(int)
    return fn

def _nn2s(make, graph_fn, X, seeds=(0, 1, 2), max_ep=300):    # same as Section 37: falls back when there is no validation period
    def fn(tr, te, yb, fold):
        last = snap_id[tr].max()
        inner = tr[snap_id[tr] <= last - 1 - GAP]; val = tr[snap_id[tr] >= last - 1]
        if len(inner) == 0:
            inner, val = tr, np.array([], dtype=int)
        s_te = np.zeros(len(te)); thr_all = []
        for sd in seeds:
            seed = 100 * fold + sd; g = graph_fn(seed)
            E, thr = max_ep // 2, 0.5
            if len(val) > 0 and len(np.unique(yb[val])) == 2:
                _, hist = _train(make, X, inner, yb, seed, g, max_ep, score_rows=val)
                auc = np.convolve(np.array([roc_auc_score(yb[val], h) for h in hist]), np.ones(5) / 5, mode="same")
                E = int(np.argmax(auc)) + 1
                cand = np.quantile(hist[E - 1], np.linspace(0.3, 0.8, 26))
                thr = float(cand[int(np.argmax([balanced_accuracy_score(yb[val], (hist[E - 1] > c).astype(int)) for c in cand]))])
            full, _ = _train(make, X, tr, yb, seed, g, E)
            s_te += full[te] / len(seeds); thr_all.append(thr)
        t = float(np.mean(thr_all))
        return s_te, (s_te > t).astype(int)
    return fn

def _bt(mode):                                                # target for every row, eligibility, persistence score, gap
    H = 28 if mode.endswith("28") else 14
    fS = np.full((n_snap, NP), np.nan); fD = np.full((n_snap, NP), np.nan); BLn = np.full((n_snap, NP), np.nan)
    cumS = np.cumsum(piv["Sales"] - piv["Delivery"], axis=0)
    for i, t in enumerate(starts):
        if t + H <= T:
            fS[i] = piv["Sales"][t:t + H].sum(0); fD[i] = piv["Delivery"][t:t + H].sum(0); BLn[i] = cumS[t + H - 1]
    ok = (~np.isnan(fS)) & (np.nan_to_num(fS) >= MIN_FUTURE_SALES_H * H / 14)
    wk_ = np.stack([piv["Sales"][t - LOOKBACK:t].sum(0) for t in starts]) / (LOOKBACK / 7)
    if mode.startswith("shortfall"):
        tgt, pers = (fS - fD) / (fS + eps), Xt[..., COL_LB]
    else:
        tgt, pers = np.clip(BLn / (wk_ + eps), -10, 10), BLW
    return np.nan_to_num(tgt).reshape(-1), ok.reshape(-1), pers.reshape(-1), int(np.ceil(H / STRIDE))

def _mf(ok, gap, n_folds=3):
    test_len = max(3, n_snap // (n_folds + 1)); out = []
    for k in range(n_folds):
        te_start = n_snap - (n_folds - k) * test_len
        tr_s, te_s = np.arange(0, te_start - gap), np.arange(te_start, te_start + test_len)
        out.append((np.where(np.isin(snap_id, tr_s) & ok)[0], np.where(np.isin(snap_id, te_s) & ok)[0]))
    return out

# %%
PAIRS = [("GCN, similarity graph", "No-graph network"), ("GCN, relation graph", "No-graph network"),
         ("GCN, relation graph", "GCN, permuted graph"), ("GCN, similarity graph", "GCN, permuted graph"),
         ("GCN, similarity graph", "Persistence"), ("Random Forest", "Persistence"),
         ("GCN, similarity graph (no degree)", "No-graph network (no degree)")]

def run_target(mode, n_boot=1000):
    global GAP
    gap_old = GAP
    tgt, ok, pers, gap = _bt(mode); GAP = gap
    models = {"Majority class": None, "Persistence": None,
              "Logistic regression": _lr(Xfin), "Random Forest": _rf(Xfin), "XGBoost": _xgb(Xfin),
              "No-graph network": _nn2s(mk_nog, lambda sd: None, Xfin),
              "GCN, relation graph": _nn2s(mk_gcn, lambda sd: EI_UNION, Xfin),
              "GCN, similarity graph": _nn2s(mk_gcn, lambda sd: EI_KNN, Xfin),
              "GCN, Plant + SubGroup": _nn2s(mk_gcn, lambda sd: EI_PS_UNION, Xfin),
              "GCN, permuted graph": _nn2s(mk_gcn, perm_graph, Xfin),
              "Random Forest (no degree)": _rf(Xnd), "No-graph network (no degree)": _nn2s(mk_nog, lambda sd: None, Xnd),
              "GCN, similarity graph (no degree)": _nn2s(mk_gcn, lambda sd: EI_KNN, Xnd)}
    n = len(tgt); pool_p = {m: np.full(n, -1) for m in models}; pool_y = np.full(n, -1); rows = []
    for k, (tr, te) in enumerate(_mf(ok, gap)):
        if len(tr) == 0:
            continue
        yb = (tgt > np.quantile(tgt[tr], 2 / 3)).astype(int)          # threshold from the training period only
        pool_y[te] = yb[te]
        sp, tp = pers[te], np.quantile(pers[tr], 2 / 3)
        for m, fn in models.items():
            if m == "Majority class": s, p = np.zeros(len(te)), np.zeros(len(te), dtype=int)
            elif m == "Persistence": s, p = sp, (sp > tp).astype(int)
            else: s, p = fn(tr, te, yb, k)
            pool_p[m][te] = p
            r = metrics_full(yb[te], s, p, tgt[te]); r.update(Target=mode, Model=m, Fold=k); rows.append(r)
        print(mode, "fold", k, "done")
    GAP = gap_old
    rows_all = np.where(pool_y >= 0)[0]; snaps = np.unique(snap_id[rows_all])
    by = {s_: rows_all[snap_id[rows_all] == s_] for s_ in snaps}
    rng = np.random.RandomState(0)
    samples_ = [np.concatenate([by[s_] for s_ in rng.choice(snaps, len(snaps))]) for _ in range(n_boot)]
    boot = {m: np.array([balanced_accuracy_score(pool_y[ix], pool_p[m][ix]) for ix in samples_]) for m in models}
    ci = {m: (balanced_accuracy_score(pool_y[rows_all], pool_p[m][rows_all]), np.percentile(boot[m], 2.5), np.percentile(boot[m], 97.5)) for m in models}
    diffs = {(a, b): (boot[a] - boot[b]) for a, b in PAIRS}
    return pd.DataFrame(rows), ci, diffs, float(1 - pool_y[rows_all].mean()), len(rows_all)

FINAL = {}
for mode in ["shortfall14", "backlog14"]:
    FINAL[mode] = run_target(mode)

all_rows = []
for mode, (df, ci, diffs, maj, nrows) in FINAL.items():
    g = df.groupby("Model", sort=False)[["BalAcc", "MacroF1", "AUC", "Precision", "Recall", "Acc"]].mean()
    g["Pooled BalAcc"] = [ci[m][0] for m in g.index]; g["CI low"] = [ci[m][1] for m in g.index]; g["CI high"] = [ci[m][2] for m in g.index]
    print(f"\n==== {mode}: {nrows} held-out rows | majority-class accuracy {maj:.3f} ====")
    display(g.round(3))
    print("Paired differences in balanced accuracy (95% CI); a CI that excludes 0 is a real difference")
    for (a, b), d in diffs.items():
        print(f"  {a} - {b}: {d.mean():+.3f}  [{np.percentile(d, 2.5):+.3f}, {np.percentile(d, 97.5):+.3f}]")
    g.insert(0, "Target", mode); all_rows.append(g.reset_index())
pd.concat(all_rows).round(4).to_csv("Final_Results.csv", index=False)
```

```python
# =====================================================================
# PART C  -  new cells for Untitled19 (14).ipynb
# Insert ALL of these directly AFTER the Section 38 code cell (the one that writes
# Final_Results.csv) and BEFORE the "Appendix A: legacy PCA-proxy results" cell.
# Each block between the "# ==== CELL ... ====" lines is ONE Colab cell.
# Blocks marked [MARKDOWN] go in a text cell; paste the comment lines without the "# ".
# Run order: 38A -> 38B-1 -> 38B-2 -> 38B-3 -> 38C -> 38D.
# Needs Sections 24-38 to have run first (piv, starts, folds, Xp3, FEATS3, EI_UNION, EI_KNN, ...).
# =====================================================================


# ==== CELL 1 [MARKDOWN] ====
# ## Section 38A: Risk index and Low / Medium / High labels (the abstract's composite risk score, leak-proof)
# The abstract promises a composite risk score cut into Low / Medium / High. The original PCA score (Part A) was
# computed from the same window as the node features, and its sign made busy products look "risky". Here:
# 1. The six operational indicators are computed on the FUTURE window [t, t+14) only; node features still come
#    from the past window [t-28, t), so features and label never share days.
# 2. Every indicator is oriented so that HIGHER = WORSE (low efficiency and high variability / gaps = risk).
# 3. Winsorising bounds, z-score statistics and the Low/Medium/High cut points are fitted on the TRAINING rows of
#    each temporal fold and then applied to the test rows (nothing is fitted on test data).
# 4. Weights are equal (transparent; no PCA sign ambiguity).


# ==== CELL 2 [CODE]  Section 38A ====
import numpy as np, pandas as pd
from scipy.stats import spearmanr

IND_NAMES = ["ProdInefficiency", "DeliveryInefficiency", "FactoryIssueDeviation", "SalesVariability",
             "ProductionVariability", "OperationalGap"]

def indicators(day_starts, length):
    """Six risk indicators per (snapshot, product) over the window [t, t+length). Rows are snapshot-major,
    the same order as Xp / snap_id / node_id."""
    out = []
    for t in day_starts:
        S, P, Fi, D = (piv[c][t:t + length] for c in ("Sales", "Production", "FactoryIssue", "Delivery"))
        St, Pt, Ft, Dt = S.sum(0), P.sum(0), Fi.sum(0), D.sum(0)
        cv = lambda A: A.std(0) / (A.mean(0) + eps)
        out.append(np.column_stack([
            1 - Pt / (St + eps),                 # production efficiency (low efficiency -> high value)
            1 - Dt / (St + eps),                 # delivery efficiency  (low efficiency -> high value)
            np.abs(1 - Ft / (Pt + eps)),         # factory-issue ratio: deviation from 1 (issue flow vs production)
            cv(S), cv(P),
            np.abs(Pt - Dt) / (St + eps)]))
    return np.concatenate(out)

IND_F  = indicators(starts, HORIZON)                              # FUTURE window  -> defines the label
IND_LB = indicators([t - HORIZON for t in starts], HORIZON)       # PAST window of equal length -> persistence baseline

def fit_index(tr):
    """Everything is fitted on training rows only."""
    X = np.clip(IND_F[tr], -5, 5)
    lo, hi = np.percentile(X, 1, axis=0), np.percentile(X, 99, axis=0)
    Xw = np.clip(X, lo, hi); mu, sd = Xw.mean(0), Xw.std(0) + 1e-6
    st = dict(lo=lo, hi=hi, mu=mu, sd=sd)
    st["cuts"] = np.quantile(risk_index(IND_F[tr], st), [1 / 3, 2 / 3])
    return st

def risk_index(ind, st):
    Xw = np.clip(np.clip(ind, -5, 5), st["lo"], st["hi"])
    return ((Xw - st["mu"]) / st["sd"]).mean(1)

def to_class(ri, st):                                              # 0 = Low, 1 = Medium, 2 = High
    return np.digitize(ri, st["cuts"])

CLASS_NAMES = ["Low", "Medium", "High"]

# ---- sanity checks (look at these printouts before trusting the label) ----
tr0, te0 = folds[0]
st0 = fit_index(tr0); ri0 = risk_index(IND_F, st0)
print("Fold-0 cut points (fitted on training rows only):", st0["cuts"].round(3))
print("Index vs. each indicator (Spearman, training rows) - all should be clearly positive:")
for j, n in enumerate(IND_NAMES):
    print(f"   {n:24s} {spearmanr(ri0[tr0], IND_F[tr0, j])[0]:+.2f}")
print("Index vs. log TotalSales (training rows): %+.2f  (Part A's PCA score was dominated by volume; this should be weak)"
      % spearmanr(ri0[tr0], Xp[tr0, 0])[0])
for k, (tr, te) in enumerate(folds):
    st = fit_index(tr); y3 = to_class(risk_index(IND_F, st), st)
    print(f"fold {k}: train class counts {np.bincount(y3[tr], minlength=3)} | test class counts {np.bincount(y3[te], minlength=3)}")


# ==== CELL 3 [MARKDOWN] ====
# ## Section 38B: Three-class evaluation (GCN vs Random Forest, XGBoost and controls)
# Same temporal folds, gap, training-only scaling and block-bootstrap confidence intervals as Sections 25-38.
# Metrics: accuracy, weighted precision / recall / F1, macro-F1 and balanced accuracy (the abstract's metrics).


# ==== CELL 4 [CODE]  Section 38B-1: model wrappers ====
import torch, torch.nn as nn, torch.nn.functional as F            # re-import torch's F (Spark's functions are imported as `sf` later)
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.utils.class_weight import compute_sample_weight
from sklearn.metrics import (accuracy_score, balanced_accuracy_score, f1_score, precision_recall_fscore_support)
from xgboost import XGBClassifier

def metrics3(y, pred):
    pw, rw, fw, _ = precision_recall_fscore_support(y, pred, average="weighted", zero_division=0)
    return {"Acc": accuracy_score(y, pred), "Prec": pw, "Rec": rw, "F1": fw,
            "MacroF1": f1_score(y, pred, average="macro", zero_division=0), "BalAcc": balanced_accuracy_score(y, pred)}

def _onehot(c):
    P = np.zeros((len(c), 3)); P[np.arange(len(c)), c] = 1.0; return P

def _full_proba(model, X):
    P = np.zeros((len(X), 3)); P[:, model.classes_.astype(int)] = model.predict_proba(X); return P

def m_majority(tr, te, y3, fold):
    return _onehot(np.full(len(te), np.bincount(y3[tr], minlength=3).argmax()))

def m_persistence(tr, te, y3, fold):                               # "the next window looks like the last one"
    st = fit_index(tr); return _onehot(to_class(risk_index(IND_LB[te], st), st))

def tab_rf(X):
    def fn(tr, te, y3, fold):
        m = RandomForestClassifier(n_estimators=300, min_samples_leaf=3, class_weight="balanced_subsample",
                                   random_state=fold, n_jobs=-1).fit(X[tr], y3[tr])
        return _full_proba(m, X[te])
    return fn

def tab_xgb(X):
    def fn(tr, te, y3, fold):
        m = XGBClassifier(n_estimators=300, max_depth=3, learning_rate=0.05, subsample=0.8, colsample_bytree=0.8,
                          objective="multi:softprob", eval_metric="mlogloss", random_state=fold, n_jobs=-1)
        m.fit(X[tr], y3[tr], sample_weight=compute_sample_weight("balanced", y3[tr]))
        return _full_proba(m, X[te])
    return fn

class GCN3(GCNv2):                                                 # GCNv2 from Section 32 with a 3-class head
    def __init__(self, d, layers=1):
        super().__init__(d, layers=layers); self.o = nn.Linear(32, 3)

class GCNrel3(GCNrel):                                             # per-relation GCN from Section 35 with a 3-class head
    def __init__(self, d):
        super().__init__(d); self.o = nn.Linear(32, 3)

def fit3(make, X, rows, y3, seed, graph, n_ep, score_rows=None):
    torch.manual_seed(seed)
    sc = StandardScaler().fit(X[rows])                             # scaler sees training rows only
    Xs = torch.tensor(sc.transform(X), dtype=torch.float)
    y = torch.tensor(np.where(y3 >= 0, y3, 0), dtype=torch.long); rt = torch.tensor(rows, dtype=torch.long)
    cnt = np.bincount(y3[rows], minlength=3).astype(float)
    w = torch.tensor(cnt.sum() / (3 * np.maximum(cnt, 1)), dtype=torch.float)
    m = make(Xs.shape[1]); opt = torch.optim.Adam(m.parameters(), lr=0.01, weight_decay=5e-4)
    hist = []
    for _ in range(n_ep):
        m.train(); opt.zero_grad()
        F.cross_entropy(m(Xs, graph)[rt], y[rt], weight=w).backward(); opt.step()
        if score_rows is not None:
            m.eval()
            with torch.no_grad():
                hist.append(torch.softmax(m(Xs, graph), 1).numpy()[score_rows])
    m.eval()
    with torch.no_grad():
        full = torch.softmax(m(Xs, graph), 1).numpy()
    return m, sc, Xs, full, hist

def nn3(make, graph_fn, X, seeds=(0, 1, 2), max_ep=300):
    """Epoch count chosen on the last two TRAINING snapshots (never the test fold), then retrained on all training rows."""
    def fn(tr, te, y3, fold):
        last = snap_id[tr].max()
        inner = tr[snap_id[tr] <= last - 1 - GAP]
        val = tr[snap_id[tr] >= last - 1]
        P = np.zeros((len(te), 3))
        for sd in seeds:
            seed = 100 * fold + sd; g = graph_fn(seed)
            _, _, _, _, hist = fit3(make, X, inner, y3, seed, g, max_ep, score_rows=val)
            f1s = np.array([f1_score(y3[val], h.argmax(1), average="macro") for h in hist])
            E = int(np.argmax(np.convolve(f1s, np.ones(5) / 5, mode="same"))) + 1
            _, _, _, full, _ = fit3(make, X, tr, y3, seed, g, E)
            P += full[te] / len(seeds)
        return P
    return fn


# ==== CELL 5 [CODE]  Section 38B-2: run the temporal cross-validation ====
INCLUDE_NN = True            # set False for a quick tabular-only test; the full run takes roughly 20-40 minutes on Colab CPU
SEEDS = (0, 1, 2)            # use (0,) for a fast check

PRED3C = {"Majority class": m_majority, "Persistence (last window)": m_persistence,
          "Random Forest": tab_rf(Xp3), "XGBoost": tab_xgb(Xp3)}
if INCLUDE_NN:
    PRED3C.update({
        "No-graph network":            nn3(lambda d: GCN3(d, layers=0), lambda sd: None, Xp3, SEEDS),
        "GCN, relation graph":         nn3(lambda d: GCN3(d, layers=1), lambda sd: EI_UNION, Xp3, SEEDS),
        "GCN, per-relation convs":     nn3(lambda d: GCNrel3(d), lambda sd: EIS_REL, Xp3, SEEDS),
        "GCN, similarity graph":       nn3(lambda d: GCN3(d, layers=1), lambda sd: EI_KNN, Xp3, SEEDS),
        "GCN, permuted graph":         nn3(lambda d: GCN3(d, layers=1), perm_graph, Xp3, SEEDS)})

n_rows = len(Xp3)
pool_y3 = np.full(n_rows, -1); pool_pred = {m: np.full(n_rows, -1) for m in PRED3C}
FOLD_STATS, fm3c = {}, {m: [] for m in PRED3C}
for k, (tr, te) in enumerate(folds):
    st = fit_index(tr); FOLD_STATS[k] = st
    y3 = np.where(elig, to_class(risk_index(IND_F, st), st), -1)   # labels for the fold, cut points from training rows only
    pool_y3[te] = y3[te]
    for m, fn in PRED3C.items():
        P = fn(tr, te, y3, k); pred = P.argmax(1)
        pool_pred[m][te] = pred; fm3c[m].append(metrics3(y3[te], pred))
    print("fold", k, "done")

rows_all = np.where(pool_y3 >= 0)[0]
print(f"\nHeld-out rows: {len(rows_all)} | class counts (Low/Medium/High): {np.bincount(pool_y3[rows_all], minlength=3)} | chance accuracy about 0.33")

# ---- block bootstrap over test snapshots (same scheme as Section 25) ----
rng = np.random.RandomState(0); B = 1000
snaps = np.unique(snap_id[rows_all]); by_snap = {s: rows_all[snap_id[rows_all] == s] for s in snaps}
samples = [np.concatenate([by_snap[s] for s in rng.choice(snaps, len(snaps))]) for _ in range(B)]
boot = {m: np.array([f1_score(pool_y3[ix], pool_pred[m][ix], average="macro", zero_division=0) for ix in samples]) for m in PRED3C}

rows = []
for m in PRED3C:
    g = metrics3(pool_y3[rows_all], pool_pred[m][rows_all])
    g.update({"Model": m, "MacroF1 CI low": np.percentile(boot[m], 2.5), "MacroF1 CI high": np.percentile(boot[m], 97.5)})
    rows.append(g)
results3 = pd.DataFrame(rows).set_index("Model")[["Acc", "Prec", "Rec", "F1", "MacroF1", "MacroF1 CI low", "MacroF1 CI high", "BalAcc"]]
display(results3.round(3))

print("\nPaired differences in macro-F1 (95% CI); a CI that excludes 0 is a real difference")
pairs = [("GCN, relation graph", "No-graph network"), ("GCN, relation graph", "GCN, permuted graph"),
         ("GCN, similarity graph", "No-graph network"), ("GCN, relation graph", "Random Forest"),
         ("GCN, relation graph", "XGBoost"), ("Random Forest", "Persistence (last window)")]
for a_, b_ in pairs:
    if a_ in boot and b_ in boot:
        d = boot[a_] - boot[b_]
        print(f"  {a_} - {b_}: {d.mean():+.3f}  [{np.percentile(d, 2.5):+.3f}, {np.percentile(d, 97.5):+.3f}]")
results3.round(4).to_csv("Final_Results_3class.csv")


# ==== CELL 6 [MARKDOWN] ====
# ## Section 38C: GNNExplainer on the final 3-class model, with human-readable reasons
# The abstract promises explanations "converted into human-readable reasons". Here GNNExplainer is run on the
# 3-class GCN trained on the last temporal fold. For each product predicted High in the latest snapshot, the
# feature mask and edge mask are translated into the abstract's reason phrases. A feature is only turned into a
# reason when its value is on the risky side of the training average (z-score), so a reason is never stated for a
# feature that is actually in a healthy range. Stability and masking checks are reported.


# ==== CELL 7 [CODE]  Section 38C ====
import re
from itertools import combinations
from torch_geometric.explain import Explainer
from torch_geometric.explain.algorithm import GNNExplainer

EXPLAIN_GRAPH = "relation"                      # "relation" (plant/storage/group edges, as in the abstract) or "similarity"
snap_edges = (lambda s: edge_index) if EXPLAIN_GRAPH == "relation" else (lambda s: EIS_SIM_LIST[s] - s * NP)
if EXPLAIN_GRAPH == "similarity":
    EIS_SIM_LIST = eis                          # list built in Section 34 (one tensor per snapshot, offset by s * NP)

tr_x, te_x = folds[-1]; st_x = FOLD_STATS[len(folds) - 1]
y3_x = np.where(elig, to_class(risk_index(IND_F, st_x), st_x), -1)
graph_train = EI_UNION if EXPLAIN_GRAPH == "relation" else EI_KNN
model_x, sc_x, Xs_x, full_x, _ = fit3(lambda d: GCN3(d, layers=1), Xp3, tr_x, y3_x, 42, graph_train, 150)
pred_x = full_x.argmax(1)
print("Explained model: held-out accuracy %.3f | macro-F1 %.3f"
      % (accuracy_score(y3_x[te_x], pred_x[te_x]), f1_score(y3_x[te_x], pred_x[te_x], average="macro")))

explainer_x = Explainer(model=model_x, algorithm=GNNExplainer(epochs=200), explanation_type="model",
                        node_mask_type="attributes", edge_mask_type="object",
                        model_config=dict(mode="multiclass_classification", task_level="node", return_type="raw"))
snap_x = lambda s: Xs_x[s * NP:(s + 1) * NP]

# feature -> (reason phrase, direction in which the feature is risky: +1 = high is bad, -1 = low is bad)
REASON = {"SalesCV": ("unstable sales demand", +1), "ProductionCV": ("unstable production", +1),
          "ProductionEfficiency": ("production inefficiency", -1), "ProductionGap": ("production below demand", -1),
          "DeliveryEfficiency": ("delivery-related problems (low delivery efficiency)", -1),
          "Shortfall7d": ("delivery-related problems (recent shortfall)", +1),
          "ShortfallLookback": ("delivery-related problems (shortfall)", +1),
          "PrevShortfall": ("delivery-related problems (earlier shortfall)", +1),
          "ShortfallTrend": ("delivery shortfall getting worse", +1), "BacklogWeeks": ("delivery-related problems (open-order backlog)", +1),
          "SalesGap": ("sales not delivered", +1),
          "FactoryIssueRatio": ("high factory issues", +1), "TotalFactoryIssue": ("high factory issues", +1),
          "GraphDegree": ("heavily connected product (network exposure)", +1)}
strip_prefix = lambda n: re.sub(r"^(lag_|chg_|d_)", "", n)

def reasons_for(row, mask, z_min=0.5, k=3):
    out = []
    for j in np.argsort(-mask):
        n = strip_prefix(FEATS3[j])
        if n in REASON and REASON[n][1] * Xs_x[row, j].item() > z_min and REASON[n][0] not in out:
            out.append(REASON[n][0])
        if len(out) == k: break
    return out if out else ["no single dominant operational driver (prediction relies on network context)"]

s_last = n_snap - 1
cases = [r for r in te_x if snap_id[r] == s_last and pred_x[r] == 2]          # predicted High in the latest snapshot
records = []
for r in cases:
    n = int(node_id[r]); ei_s = snap_edges(s_last)
    ex = ex = explainer_x(snap_x(s_last), ei_s, index=n)
    nm = ex.node_mask[n].detach().numpy()
    inc = (ei_s[1] == n).nonzero().view(-1)                                  # edges that point into the explained product
    nb = {}
    for e in inc[torch.argsort(-ex.edge_mask[inc])][:12].tolist():
        nb.setdefault(prods[int(ei_s[0, e])], float(ex.edge_mask[e]))
    records.append({"Product": prods[n], "Predicted": CLASS_NAMES[pred_x[r]], "P(High)": round(float(full_x[r, 2]), 3),
                    "Reasons": "; ".join(reasons_for(r, nm)), "Most influential connected products": ", ".join(list(nb)[:3])})
explanations = pd.DataFrame(records)
display(explanations)
explanations.to_csv("Explainable_Risk_Predictions_final.csv", index=False)

# ---- stability: 5 explainer runs with different seeds on one case ----
if cases:
    r0 = cases[0]; n0 = int(node_id[r0]); masks = []
    for sd in range(5):
        torch.manual_seed(sd); masks.append(explainer_x(snap_x(s_last), snap_edges(s_last), index=n0).node_mask[n0].detach().numpy())
    masks = np.array(masks)
    print("Mean Spearman between explainer runs: %.3f" % np.mean([spearmanr(masks[a], masks[b])[0] for a, b in combinations(range(5), 2)]))

# ---- fidelity: replace the top-3 features of the product by the training mean (0 after scaling) and see if the class changes ----
FID_SNAPS = 2; rng_f = np.random.RandomState(0); ch_top = ch_rnd = tot = 0
for r in [r for r in te_x if snap_id[r] >= n_snap - FID_SNAPS]:
    s, n = int(snap_id[r]), int(node_id[r]); x0 = snap_x(s); ei_s = snap_edges(s)
    nm = explainer_x(x0, ei_s, index=n).node_mask[n].detach().numpy()
    with torch.no_grad():
        c0 = model_x(x0, ei_s)[n].argmax().item()
        for idx, name in ((np.argsort(-nm)[:3], "top"), (rng_f.choice(len(nm), 3, replace=False), "rnd")):
            xm = x0.clone(); xm[n, torch.as_tensor(np.asarray(idx), dtype=torch.long)] = 0.0
            changed = int(model_x(xm, ei_s)[n].argmax().item() != c0)
            ch_top += changed if name == "top" else 0; ch_rnd += changed if name == "rnd" else 0
    tot += 1
print(f"Prediction changed when masking the top-3 explained features: {ch_top}/{tot} | random 3 features: {ch_rnd}/{tot}")


# ==== CELL 8 [MARKDOWN] ====
# ## Section 38D: PySpark window features (big-data processing in the final pipeline)
# The rolling-window features of Section 24 are recomputed with PySpark (range join of a snapshot table to the daily
# records, then group-by aggregation) and compared with the NumPy features that the models actually use. A small
# replication test shows the same code running on 1x, 10x and 50x more products. This is a single-machine Spark
# # demonstration of a distributed-style pipeline, not a cluster benchmark.


# # ==== CELL 9 [CODE]  Section 38D ====
# import time, functools
# from pyspark.sql import SparkSession, DataFrame
# from pyspark.sql import functions as sf                      # imported as `sf` so torch's F is not overwritten

# spark = SparkSession.builder.appName("SupplyChainRiskBigDataProcessing").getOrCreate()
# date_to_day = {pd.Timestamp(d): i for i, d in enumerate(dates_all)}
# long_df = md.assign(day=md["Date"].map(date_to_day))[["Product", "day", "Sales", "Production", "FactoryIssue", "Delivery"]]
# sdf = spark.createDataFrame(long_df)
# snap_tbl = spark.createDataFrame(pd.DataFrame({"snap": np.arange(n_snap), "t": np.array(starts, dtype=int)}))

# def spark_window_features(sdf):
#     win = sdf.join(sf.broadcast(snap_tbl), (sdf.day >= snap_tbl.t - LOOKBACK) & (sdf.day < snap_tbl.t))
#     return (win.groupBy("snap", "Product").agg(
#                 sf.sum("Sales").alias("S_t"), sf.sum("Production").alias("P_t"), sf.sum("FactoryIssue").alias("F_t"),
#                 sf.sum("Delivery").alias("D_t"), sf.stddev_pop("Sales").alias("sd_S"), sf.avg("Sales").alias("mu_S"),
#                 sf.stddev_pop("Production").alias("sd_P"), sf.avg("Production").alias("mu_P"))
#             .withColumn("DeliveryEfficiency", sf.col("D_t") / (sf.col("S_t") + eps))
#             .withColumn("ProductionEfficiency", sf.col("P_t") / (sf.col("S_t") + eps))
#             .withColumn("SalesCV", sf.col("sd_S") / (sf.col("mu_S") + eps))
#             .withColumn("ProductionCV", sf.col("sd_P") / (sf.col("mu_P") + eps)))

# feat_sp = spark_window_features(sdf).toPandas()
# feat_sp["pi"] = feat_sp["Product"].map({p: i for i, p in enumerate(prods)})
# def _grid(col):
#     A = np.full((n_snap, NP), np.nan); A[feat_sp["snap"].values, feat_sp["pi"].values] = feat_sp[col].values; return A
# # column indices in Xraw (Section 24): 0 TotalSales, 4 SalesCV, 5 ProductionCV, 6 ProductionEfficiency, 7 DeliveryEfficiency
# checks = {"S_t": 0, "SalesCV": 4, "ProductionCV": 5, "ProductionEfficiency": 6, "DeliveryEfficiency": 7}
# print("PySpark vs NumPy window features (max absolute difference over all snapshot-product cells that Spark returned):")
# for col, j in checks.items():
#     A = _grid(col); ok = ~np.isnan(A)
#     print(f"   {col:22s} max |diff| = {np.nanmax(np.abs(A[ok] - Xraw[..., j][ok])):.2e}   (cells compared: {int(ok.sum())} of {n_snap * NP})")

# print("\nScaling demonstration (same Spark code on replicated products):")
# for k in (1, 10, 50):
#     big = functools.reduce(DataFrame.unionByName,
#                            [sdf.withColumn("Product", sf.concat(sf.col("Product"), sf.lit(f"_{j}"))) for j in range(k)])
#     t0 = time.time(); n_out = spark_window_features(big).count(); dt = time.time() - t0
#     print(f"   {k:3d}x -> {k * md['Product'].nunique():5d} products, {len(long_df) * k:8d} daily records, {n_out:7d} feature rows in {dt:5.1f} s")
```

```python
import time
from pyspark.sql import SparkSession
from pyspark.sql import functions as sf

spark = SparkSession.builder.appName("SupplyChainRiskBigDataProcessing") \
        .config("spark.sql.shuffle.partitions", "8").getOrCreate()

# FIX: rebuild the (snapshots x products x features) tensor; the Robustness cell overwrote the global Xraw
Xsn = np.stack([snap_features(t) for t in starts])

date_to_day = {pd.Timestamp(d): i for i, d in enumerate(dates_all)}
long_df = md.assign(day=md["Date"].map(date_to_day))[["Product", "day", "Sales", "Production", "FactoryIssue", "Delivery"]]
sdf = spark.createDataFrame(long_df)
snap_tbl = spark.createDataFrame(pd.DataFrame({"snap": np.arange(n_snap), "t": np.array(starts, dtype=int)}))

def spark_window_features(s):
    win = s.join(sf.broadcast(snap_tbl), (s.day >= snap_tbl.t - LOOKBACK) & (s.day < snap_tbl.t))
    return (win.groupBy("snap", "Product").agg(
                sf.sum("Sales").alias("S_t"), sf.sum("Production").alias("P_t"), sf.sum("FactoryIssue").alias("F_t"),
                sf.sum("Delivery").alias("D_t"), sf.stddev_pop("Sales").alias("sd_S"), sf.avg("Sales").alias("mu_S"),
                sf.stddev_pop("Production").alias("sd_P"), sf.avg("Production").alias("mu_P"))
            .withColumn("DeliveryEfficiency", sf.col("D_t") / (sf.col("S_t") + eps))
            .withColumn("ProductionEfficiency", sf.col("P_t") / (sf.col("S_t") + eps))
            .withColumn("SalesCV", sf.col("sd_S") / (sf.col("mu_S") + eps))
            .withColumn("ProductionCV", sf.col("sd_P") / (sf.col("mu_P") + eps)))

feat_sp = spark_window_features(sdf).toPandas()
feat_sp["pi"] = feat_sp["Product"].map({p: i for i, p in enumerate(prods)})
def _grid(col):
    A = np.full((n_snap, NP), np.nan); A[feat_sp["snap"].values, feat_sp["pi"].values] = feat_sp[col].values; return A

checks = {"S_t": 0, "SalesCV": 4, "ProductionCV": 5, "ProductionEfficiency": 6, "DeliveryEfficiency": 7}
print("PySpark vs NumPy window features (max absolute difference):")
for col, j in checks.items():
    A = _grid(col); ok = ~np.isnan(A)
    print(f"   {col:22s} max |diff| = {np.nanmax(np.abs(A[ok] - Xsn[..., j][ok])):.2e}   (cells compared: {int(ok.sum())} of {n_snap * NP})")

print("\nScaling demonstration (same Spark code on replicated products):")
for k in (1, 10, 50):
    rep = spark.range(k).withColumnRenamed("id", "rep")          # replicate with a cross join (no 50-deep union chain)
    big = (sdf.crossJoin(rep)
              .withColumn("Product", sf.concat_ws("_", sf.col("Product"), sf.col("rep").cast("string"))).drop("rep"))
    t0 = time.time(); n_out = spark_window_features(big).count(); dt = time.time() - t0
    print(f"   {k:3d}x -> {k * md['Product'].nunique():5d} products, {len(long_df) * k:8d} daily records, {n_out:7d} feature rows in {dt:5.1f} s")
```
